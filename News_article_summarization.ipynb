```python
# ============================================================
# NEWS ARTICLE SUMMARIZATION USING MACHINE LEARNING
# Google Colab - Single Cell Complete Code
# ============================================================

# Install Required Libraries
!pip -q install transformers sentencepiece accelerate

# Import Libraries
import torch
from transformers import pipeline

print("=" * 70)
print("      NEWS ARTICLE SUMMARIZATION USING MACHINE LEARNING")
print("=" * 70)

# Check GPU
device = 0 if torch.cuda.is_available() else -1

if device == 0:
    print("GPU Status : Available")
else:
    print("GPU Status : Not Available (Using CPU)")

print("\nLoading BART Model... Please wait.\n")

# Load Pre-trained BART Model
summarizer = pipeline(
    "summarization",
    model="facebook/bart-large-cnn",
    tokenizer="facebook/bart-large-cnn",
    device=device
)

print("Model Loaded Successfully!")

# ------------------------------------------------------------
# Sample News Article
# ------------------------------------------------------------

news_article = """
Artificial Intelligence (AI) is transforming industries across the globe.
In healthcare, AI helps doctors detect diseases earlier and more accurately.
Banks use AI systems to identify fraudulent transactions, while educational
institutions are using AI-powered platforms to provide personalized learning.

Researchers believe AI will continue to grow rapidly in the coming years.
However, experts also highlight concerns such as privacy, bias,
misinformation, and job displacement. Governments and technology companies
are working together to develop ethical AI policies that ensure responsible
innovation.

Many companies are investing billions of dollars into AI research and
development, making Artificial Intelligence one of the fastest-growing
fields in computer science.
"""

print("\n" + "=" * 70)
print("ORIGINAL ARTICLE")
print("=" * 70)
print(news_article)

print("\nGenerating Summary...\n")

summary = summarizer(
    news_article,
    max_length=80,
    min_length=30,
    do_sample=False
)

print("=" * 70)
print("SUMMARY")
print("=" * 70)
print(summary[0]["summary_text"])

# ------------------------------------------------------------
# User Input
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SUMMARIZE YOUR OWN ARTICLE")
print("=" * 70)

choice = input("Do you want to summarize your own article? (yes/no): ").strip().lower()

if choice == "yes":
    article = input("\nPaste your news article below:\n\n")

    user_summary = summarizer(
        article,
        max_length=80,
        min_length=30,
        do_sample=False
    )

    print("\n" + "=" * 70)
    print("YOUR SUMMARY")
    print("=" * 70)
    print(user_summary[0]["summary_text"])

print("\nProject Completed Successfully!")
```
